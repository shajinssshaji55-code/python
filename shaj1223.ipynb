```python
class Bank:
    def __init__(self):
        self.balance = 0
        print("Welcome to the Machine")

    def deposit(self):
        amount = float(input("Enter amount to be deposited: "))
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self):
        amount = float(input("Enter amount to be withdrawn: "))
        if self.balance >= amount:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display(self):
        print("Available Balance =", self.balance)


if __name__ == "__main__":
    s = Bank()
    s.deposit()
    s.withdraw()
    s.display()
```

```python
def atm():
    balance = 1000
    pin = "1234"

    print("Welcome to the ATM")

    enter_pin = input("Enter your 4-digit PIN: ")

    if enter_pin != pin:
        print("Incorrect PIN. Exiting")
        return

    while True:
        print("\nATM Menu:")
        print("1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            print(f"Your current balance is: ${balance}")

        elif choice == '2':
            amount_str = input("Enter amount to deposit: $")

            if amount_str.isdigit():
                amount = int(amount_str)

                if amount > 0:
                    balance += amount
                    print(f"Successfully deposited ${amount}")
                    print(f"New balance: ${balance}")
                else:
                    print("Deposit amount must be positive")
            else:
                print("Invalid input. Please enter a number.")

        elif choice == '3':
            amount_str = input("Enter amount to withdraw: $")

            if amount_str.isdigit():
                amount = int(amount_str)

                if amount > 0:
                    if amount <= balance:
                        balance -= amount
                        print(f"Successfully withdrew ${amount}")
                        print(f"Remaining balance: ${balance}")
                    else:
                        print("Insufficient funds.")
                else:
                    print("Withdrawal amount must be positive")
            else:
                print("Invalid input. Please enter a number.")

        elif choice == '4':
            print("Thank you for using the ATM. Goodbye!")
            break

        else:
            print("Invalid choice. Please select a valid option (1-4)")


atm()
```

```python
from abc import ABC,abstractmethod

class Dog(ABC):
    
    def __init__(self,name):
        self.name = name
        
    @abstractmethod
    def sound(self):
        pass
    def display_name(self):
        print(f"Dog's Name: {self.name}")

class Labrador(Dog):
    def sound(self):
        print("Labrador woof!")

class Beagle(Dog):
    def sound(self):
        print("Beagle barks!")

dogs=[Labrador("buddy"),Beagle("Charly")]

for dog in dogs:
    dog.display_name()
    dog.sound()
        
    
```

```python
class Maths:

    def __init__(self, length, breadth, height):
        self.length = length
        self.breadth = breadth
        self.height = height

    def area(self):
        return 2 * (
            self.length * self.breadth +
            self.breadth * self.height +
            self.height * self.length
        )

    def volume(self):
        return self.length * self.breadth * self.height


maths2 = Maths(2, 4, 6)

print("Surface area:", maths2.area())
print("Volume:", maths2.volume())
```
