```python
from sklearn.linear_model import LinearRegression
import numpy as np
import joblib
x = np.array([[1000,2],[1500,3],[2000,4]])
y = np.array([200000,300000,400000])

model = LinearRegression()
model.fit(x,y)

import joblib
joblib.dump(model,"model.pkl")
loaded_model =joblib.load("model.pkl")
prediction = loaded_model.predict([[1200,2]])
print(prediction)
```

```python
import pickle
with open('book9','wb')as file:
    pickle.dump(model,file)
with open('book9','rb')as file:
    mp=pickle.load(file)
mp.coef_
mp.intercept_
mp.predict([[3100,4]])
```

```python
import pandas as pd
df=pd.read_csv('linear6.csv')
df              
```

```python
dummies = pd.get_dummies(df.town).astype(int)
dummies
```

```python
df_dummies = pd.concat([df,dummies],axis = 'columns')
df_dummies
```

```python
df_dummies.drop('town',axis='columns',inplace=True)
df_dummies
```

```python
df_dummies.drop('west windsor',axis='columns',inplace=True)
df_dummies
```

```python
x = df_dummies.drop('price', axis='columns')
y = df_dummies.price
```

```python
from sklearn import linear_model
model = linear_model.LinearRegression()
model.fit(x,y)
```

```python
model.predict(x)
model.score(x,y)
```
